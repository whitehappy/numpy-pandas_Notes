```python
#绘制折线图
import matplotlib.pyplot as plt
from matplotlib import rcParams #这个rcParams属性用于控制字体
#rcParams['font.family'] = 'SimHei' #matplotlib绘图默认字体不显示中文，需要修改属性来进行中文显示,设置这个参数可以
rcParams['font.sans-serif'] = 'SimHei' #matplotlib绘图默认字体不显示中文，需要修改属性来进行中文显示，设置这个参数也可以修改字体
#创建图表，设置大小
plt.figure(figsize=(10,5))#画折线图的纸的大小。画布的概念

#绘制的数据
month = ['1月','2月','3月','4月']
sales = [100,150,80,130]

#绘制折线图
plt.plot(month,sales,
         label='产品A',
        color='orange',
        linewidth=3,
        linestyle='--',
        marker='o')

#添加标题
plt.title('2025年销售趋势',color='red',fontsize=20)
#添加坐标轴的标签
plt.xlabel('月份',fontsize=10)
plt.ylabel('销售额（万元）',fontsize=10)
#添加图例（图例含义：折线图中的这条蓝色线代表的是谁）
plt.legend(loc='upper left')
#添加网格线
plt.grid(True,alpha=0.1,color='blue',linestyle='--') #alpha网格线透明度控制,linestyle控制网格线风格（'-'实线，'--'虚线）
#plt.grid(axis='x')
#设置刻度的字体大小
plt.xticks(rotation=0,fontsize=8)#rotation控制坐标轴上刻度的旋转角度(用于字体较密集时使用)，fontsize控制坐标轴上刻度的字体大小
plt.yticks(rotation=0,fontsize=8)
#设置y轴范围
plt.ylim(0,160)
#在每个数据点上显示数值
for x,y in zip(month,sales):
    print(x,y)
    plt.text(x,y+4,str(y),ha='center',va='bottom',fontsize=10)

#显示图表
plt.show()
```

```python
#绘制柱状图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif'] = 'SimHei'

#创建图表绘制大小
plt.figure(figsize=(10,5))
# 要绘图的数据
subjects = ['语文','数学','英语','科学']
scores = [85,92,78,88]

#绘制柱状图
plt.bar(subjects,scores,
        label=['语文','数学','英语','科学'],
       color=['red','blue','orange','green'],
       width=0.5)
#添加标题
plt.title('小红成绩分布',color = 'red',fontsize=20)
#添加坐标轴标签
plt.xlabel('科目',fontsize=10)
plt.ylabel('成绩',fontsize=10)
#添加图例
plt.legend(loc='upper left')
#添加网格线
plt.grid(axis='y',alpha=0.1,color='black',linestyle='--')
#设置刻度字体大小
plt.xticks(rotation = 0,fontsize=12)
plt.yticks(rotation = 0,fontsize=12)
#设置y轴范围
plt.ylim(0,100)
#数据点的显示
for x,y in zip(subjects,scores):
    plt.text(x,y,str(y),ha="center",va="bottom",fontsize=10)
#自动优化排版
plt.tight_layout()


#展示条形图
plt.show()
```

```python
#绘制条形图
import matplotlib.pyplot as plt
from matplotlib import rcParams

rcParams['font.family'] = 'SimHei'

#创建图标，设置大小
plt.figure(figsize=(10,5))

#要绘制的数据
countrys = ['United States','China','Japaa','Germany','India']
gdp = [92,85,42,32,28]
#绘制条形图
plt.barh(countrys,gdp,color='orange')
#添加标题
plt.title('2025年GDP排名',color='red',fontsize=20)
#添加坐标轴标签
plt.xlabel('GDP',fontsize=10)
plt.ylabel('国家',fontsize=10)
#设定x轴范围
plt.xlim(0,100)
#美观
plt.tight_layout()


#展示条形图
plt.show()
```

```python
#绘制饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

#创建图表，绘制大小
plt.figure(figsize=(10,5))

#要绘制的数据
things = ['学习','娱乐','运动','睡觉','其他'] 
times = [6,3,1,8,5]
colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499']#配色方案
#绘制饼图
plt.pie(times,labels=things,
       autopct='%.1f%%',#显示百分比
       startangle = 90,#调整初始画图角度
        colors=colors
       )
#添加标题
plt.title('一天时间分布',color='orange',fontsize=20)
#饼图中没有x轴和y轴，所以x轴和y轴标签不存在
#自动优化排版
plt.tight_layout()

#展示图标
plt.show()
```

```python
#绘制环形图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

#绘制图标大小
plt.figure(figsize=(10,5))

#设置数据
things = ['学习','娱乐','运动','睡觉','其他'] 
times = [6,3,1,8,5]
colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499']#配色方案
#绘制环形图
plt.pie(times,labels=things,
        autopct='%.1f%%',
        colors = colors,
        startangle = 90,
        wedgeprops={'width':0.25},#配置环形图圆环宽度
        pctdistance = 0.9 #设置百分比的位置
       )
#添加标题
plt.title('一天的时间分布',color = 'orange',fontsize=20)
plt.text(0,0,'总计：\n100%',ha = 'center',va='bottom',fontsize=10)
#自动优化图形
plt.tight_layout()
#展示环形图
plt.show()
```

```python
#绘制爆炸式饼图
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

#创建图表，绘制大小
plt.figure(figsize=(10,5))

#要绘制的数据
things = ['学习','娱乐','运动','睡觉','其他'] 
times = [6,3,1,8,5]
colors = ['#66b3ff','#99ff99','#ffcc99','#ff9999','#ff4499']#配色方案
explode = [0.1,0,0,0,0]#设置突出块的位置
#绘制饼图
plt.pie(times,labels=things,
       autopct='%.1f%%',#显示百分比
       startangle = 0,#调整初始画图角度
        colors=colors,
        explode = explode,#设置突出块
        shadow = True,#设置阴影
       )
#添加标题
plt.title('一天时间分布',color='orange',fontsize=20)
#饼图中没有x轴和y轴，所以x轴和y轴标签不存在
#自动优化排版
plt.tight_layout()

#展示图标
plt.show()#
```

```python
#绘制散点图
import matplotlib.pyplot as plt
import random
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

#创建图标
plt.figure(figsize=(10,8))

#绘制散点图的数据
x = []
y = []
for i in range(1000):
    tmp = random.uniform(0,10)
    x.append(tmp)
    tmp2  =2*tmp +random.gauss(0,2)
    y.append(tmp2)
#绘制散点图
plt.scatter(x,y,
           color = 'blue',#点的颜色
           alpha=0.5,#点透明度
           s=20,#圈的大小
           label = '数据')
#添加标题
plt.title('X变量与Y变量的关系',color = 'red', fontsize = 20)
plt.xlabel('x变量',fontsize = 10)
plt.ylabel('y变量',fontsize = 10)
plt.legend(loc = 'upper left')           
#添加网格线
plt.grid(True,alpha=0.1,color='blue',linestyle='--')
#设置刻度字体大小
plt.xticks(rotation=0,fontsize=12)
plt.yticks(rotation=0,fontsize=12)
#设置y轴的范围
plt.ylim(0,30)
plt.plot([0,10],[0,20],color='red',linewidth=2,linestyle='--')

#优化
plt.tight_layout()
#展示图标
plt.show()
```

```python
#分析案例：温度分析
#1.导入库
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif'] = 'SimHei'
rcParams['axes.unicode_minus'] = False

#2.导入数据
df = pd.read_csv('data/weather.csv')
df.head()

#绘制气温的趋势变化图
df['date'] = pd.to_datetime(df['date'])#把时间数据从字符串转化为pandas中的日期类型
df = df[df['date'].dt.year == 2015]
plt.figure(figsize=(10,10))
plt.plot(df['date'],df['temp_max'],label='最高气温')
plt.plot(df['date'],df['temp_min'],label='最低气温')
plt.title('2015年气温趋势变化图')
plt.xlabel('日期')
plt.ylabel('温度')
plt.legend()
plt.show()
```

```python
df['temp_mean'] = (df['temp_max'] + df['temp_min'])/2
df.head(5)
plt.figure(figsize=(10,10))
plt.plot(df['date'],df['temp_mean'],label='平均气温')
plt.title('2015年气温趋势变化图')
plt.xlabel('日期')
plt.ylabel('温度')
plt.legend()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']

penguins = pd.read_csv("data/penguins.csv")
penguins.dropna(inplace = True)
penguins.info()
penguins.head()
```

```python
#直方图
sns.histplot(data=penguins,x="species")
```

```python
#核密度估计图
'''
核密度估计图（KDE,Kernel Density Estimate Plot）是一种用于显示数据分布
的统计图表，它通过平滑直方图的方法来估计数据的概率密度函数，使得分布图
看起来更加连续和平滑。核密度估计是一种非参数方法，用于估计随机变量的概
率密度函数。其基本思想是，将每个数据点视为一个“核”（通常是高斯分布），
然后将这些核的贡献相加以形成平滑的密度曲线
'''
#绘制喙长度的核密度估计图
sns.kdeplot(data=penguins,x="bill_length_mm")
```

```python
sns.histplot(data=penguins,x="bill_length_mm",kde=True)
```

```python
'''
计数图
计数图用于绘制分类变量的计数分布图，显示每个类别在数据集中出现的次数，是分析分类数据非常直观的工具，
可以快速了解类别的分布情况
'''
#绘制不同岛屿企鹅数量的计数图
sns.countplot(data=penguins,x='island')
```

```python
sns.histplot(data=penguins,x='island')
```

```python
#散点图
#绘制横轴为体重，纵轴为脚蹼长度的散点图。可通过hue参数设置不同组别进行对比
sns.set_style("whitegrid")
sns.scatterplot(data=penguins,x="body_mass_g",
               y="flipper_length_mm",hue="sex")
```
