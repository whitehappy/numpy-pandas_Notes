```python
import numpy as np
import pandas as pd
np.random.seed(42)
#scores = pd.Series(np.random.randint(50,101,10),index = ['学生' + str(i) for i in range(1,11)])
values = np.random.randint(50,101,10)
index = []
for i in range(1,11):
    index.append('学生'+str(i))
scores = pd.Series(values,index)
print(scores)
average = scores.mean()
max = scores.max()
min = scores.min()
print("平均分为：",average)
print("最高分为：",max)
print("最低分为：",min)
over_average = scores[scores>average]
print(len(over_average))
print(over_average)
print('超越个数为',over_average.count())
```

```python
import pandas as pd
temperatures = pd.Series([28,31,29,32,30,27,33],index=['周一','周二','周三','周四','周五','周六','周日'])
over_temperature_30 = temperatures[temperatures>30]
print("超过30度的天数：",over_temperature_30.count())
mean = temperatures.mean()
print("平均温度为：",mean)
t2 = temperatures.sort_values(ascending =False)
print("从高到低排序为：\n",t2)
#找出温度变化最大的两天
#28 31 29 32 30 27 33
#none 3 -2 3 -2 -3 6
t3 = temperatures.diff().abs()#后一元素和前一个元素的差值 计算series的变化值
print("温度变化最大的两天",*(t3.sort_values(ascending=False).keys()[:2].tolist()))#*()去掉中括号 ， tolist把index对象变为列表
#t3.sort_values(ascending=False).index
```

```python
import pandas as pd
import numpy as np
sales = pd.Series([120,135,145,160,155,170,180,175,190,200,210,220],index=pd.date_range('2022-01-01',periods=12,freq='MS'))
print(sales)
```

```python
#季度的平均销量 sales
sales.resample('QS').count()
#sales.resample('YS').mean() == sales.mean() 因为样本就是12个月
sales.resample('QS').mean()#resample()重新采样 'QS'每个季度的开始
sales.resample('QS').sum()
```

```python
print('销量最高的月份',sales.idxmax())
```

```python
print('月环比的增长率：',pct_change)
pct_change = sales.pct_change()
pct_change_judge = pct_change>0
#用滑动窗口来看是否存在连续两个月增长
pct_change_judge[pct_change_judge.rolling(3).sum() == 3].keys().tolist()
```

```python
import pandas as pd
import numpy as np
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0,100,24),
                        index=pd.date_range('2025-01-01',periods=24,freq='h'))
print(hourly_sales)
#按天进行重采样计算每日总销售额
day_sales = hourly_sales.resample('D').sum()
#计算每天营业时间（8：00-22：00）和非营业时间销售额的比例
#hourly_sales.between_time('8:00', '22:00')#筛选一段时间内的series ，between_time方式计算一段时间销售额
business_hours_sales = hourly_sales[(hourly_sales.index.hour>=8) & (hourly_sales.index.hour <= 22)]#布尔索引方式计算一段时间销售额
#print(business_hours_sales.sum()/(day_sales - business_hours_sales.sum()))
not_business_hours_sales = hourly_sales.drop(business_hours_sales.index)#drop()去除原有系列中的一部分系列
print(business_hours_sales.sum()/not_business_hours_sales.sum())
#此外取反也行

#找出销售额最高的三个小时
print(hourly_sales.nlargest(3).index.tolist())
```

```python
import pandas as pd
data = {
    '姓名':['张三','李四','王五','赵六','钱七'],
    '数学':[85,92,78,88,95],
    '英语':[90,88,85,92,80],
    '物理':[75,80,88,85,90]
}

scores = pd.DataFrame(data,index=[1,2,3,4,5],columns=['姓名','数学','英语','物理'])
#计算每位学生的总分和平均分
scores['总分'] = scores[['数学','英语','物理']].sum(axis=True)
#print('张三总和为：',scores.loc[1,'数学':'物理'].sum())
#scores['平均分'] = scores[['数学','英语','物理']].mean(axis=True)
scores['平均分'] = scores['总分'] / 3
#找出数学平均分高于90分或英语成绩高于85分的学生
print(scores[(scores['数学']>90) | (scores['英语']>85)])
#按总分从高到低排序，并输出前三名学生
r1 = scores.sort_values('总分',ascending=False).head(3)
r2 = scores.nlargest(3,columns=['总分'])
print(r1)
print(r2)
```

```python
import pandas as pd

data = {
    '产品名称':['A','B','C','D'],
    '单价':[100,150,200,120],
    '销量':[50,30,20,40]
}

products = pd.DataFrame(data,index=[1,2,3,4],columns=['产品名称','单价','销量'])
products['总销售额'] = products['单价']*products['销量']
print(products.nlargest(1,columns=['总销售额']))
products.sort_values('总销售额',ascending = False)
```

```python
import pandas as pd

data = {
    '用户ID':[101,102,103,104,105],
    '用户名':['Alice','Bob','Charlie','David','Eve'],
    '商品类别':['电子产品','服饰','电子产品','家居','服饰'],
    '商品单价':[1200,300,800,150,200],
    '购买数量':[1,3,2,5,4]
}

products = pd.DataFrame(data,index=[1,2,3,4,5],columns=['用户ID','用户名','商品类别','商品单价','购买数量'])
#计算每位用户的总消费金额
products['总消费金额'] = products['商品单价']*products['购买数量']
#找出消费金额最高的用户，并输出其所有信息
products.nlargest(1,columns = ['总消费金额'])
#计算所有用户的平均消费金额
products['总消费金额'].mean()
#计算电子产品的总购买数量
products[products['商品类别'] == '电子产品']['购买数量'].sum()
```
