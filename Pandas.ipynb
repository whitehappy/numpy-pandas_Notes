```python
# 缺失值的处理
#nan:not a number(如无穷大和无穷小无法用数字表示的)
#None:不存在
#pd.NA:不存在，（倾向于没有数值）
import pandas as pd
import numpy as np

s = pd.Series([12,25,np.nan,None,pd.NA])
df = pd.DataFrame([[1,pd.NA,2],[2,3,5],[None,4,6]],columns=['第一列','第二列','第三列'])
print(s)

#查看是否是缺失值(Series中的缺失值)
print(s.isna())
print(s.isnull())

#查看是否是缺失值(DataFrame中的缺失值)
print(df.isna())
print(df.isnull())

print(df.isna().sum(axis=1))#查看缺失值个数(按照行看)
print(df.isnull().sum())#查看缺失值个数(按照列看)
print('-'*30)
#剔除缺失值
print(s.dropna())

print('-'*30)
print(df)
print(df.dropna())#在DataFrame中只要一个index行中存在缺失值,直接整条删除

#如果存在有不想删的(dataFrame中)
print(df.dropna(how='all'))#如果每行所有的值均为缺失值,才删去这一行
print(df.dropna(thresh=2))#如果每行至少有n个(这里为2)不是缺失值就保留
print(df.dropna(thresh=3))
print(df.dropna(axis=1))#剔除一整列的记录
print(df.dropna(subset=['第一列']))#如果某列有缺失值，则删除此列的该位置对应行

#填充缺失值
df = pd.read_csv('data/weather_withna.csv')
df.tail()
df.isna().sum(axis=0)
df.head()
print(df.fillna({'temp_max':20,'wind':2.5}).tail())#使用字典来填充
print(df.fillna(df[['temp_max','wind']].mean()).tail())#使用统计值来填充
print(df.ffill().tail())#用前面的相邻值进行添加front
print(df.bfill().tail())#用后面的相邻值进行填充behind
```

```python
#数据变形
import pandas as pd
data ={
    'ID':[1,2],
    'name':['alice smith','bob smith'],
    'Math':[90,85],
    'English':[88,92],
    'Science':[95,89]
}
df = pd.DataFrame(data)
#分列
df[['first','last']]=df['name'].str.split(" ",expand=True)
df = pd.read_csv('data/sleep.csv')
df.head()
df[['blood_pressure_high','blood_pressure_low']] = df['blood_pressure'].str.split("/",expand=True)

df['blood_pressure_high']=df['blood_pressure_high'].astype('int16')
df['blood_pressure_low']=df['blood_pressure_low'].astype('int16')
df.info()
df.blood_pressure_high.mean()
df.blood_pressure_low.mean()
```

```python
df.T #行列转置
df
'''
ID name  科目 成绩
1 alice Math 90
1 alice English 88
1 alice Science 95
'''

#宽表转换为长表
df2 = pd.melt(df,id_vars=['ID','name'],var_name='科目',value_name='分数')
df2.sort_values('name')
```

```python
#长表转宽表
pd.pivot(df2,index=['ID','name'],columns='分数',values='科目')
```

```python
#数据分箱 pd.cut(x,bins,labels)

import pandas as pd

df = pd.read_csv('data/employees.csv')
df.head(10)
```

```python
df1 = df.head(10)[['employee_id','salary']]
df1
```

```python
pd.cut(df1['salary'],bins=2) #bins=n ,及分成n段区间,起始值、结束值是所有数据的最小值和最大值
#4180~14100~24000
pd.cut(df1['salary'],bins=3).value_counts()
pd.cut(df1['salary'],bins=[0,10000,20000,30000]).value_counts()#bins=list,分成n段区间
df1['收入范围']=pd.cut(df1['salary'],bins=[0,10000,20000,30000],labels=['低','中','高'])
pd.qcut(df1['salary'],3).value_counts()#按人数区分划分，等频率划分
```

```python
#睡眠数据
import pandas as pd

df = pd.read_csv('data/sleep.csv')
df.head(10)

df1 = df.head(10)[['person_id','sleep_quality']]
df1
df['睡眠质量']=pd.cut(df['sleep_quality'],bins=3,labels=['差','中','优'])
df['睡眠质量'].value_counts()
df.head(10)
df['gender']=df['gender'].astype('category')
df['gender'].value_counts()
#字符串-->类别-->统计
#数值-->分箱-->统计
print(df['gender'].dtype)
print(df['睡眠质量'].dtype)
```
