```python
# 缺失值的处理
#nan:not a number(如无穷大和无穷小无法用数字表示的)
#None:不存在
#pd.NA:不存在，（倾向于没有数值）
import pandas as pd
import numpy as np

s = pd.Series([12,25,np.nan,None,pd.NA])
df = pd.DataFrame([[1,pd.NA,2],[2,3,5],[None,4,6]],columns=['第一列','第二列','第三列'])
print(s)

#查看是否是缺失值(Series中的缺失值)
print(s.isna())
print(s.isnull())

#查看是否是缺失值(DataFrame中的缺失值)
print(df.isna())
print(df.isnull())

print(df.isna().sum(axis=1))#查看缺失值个数(按照行看)
print(df.isnull().sum())#查看缺失值个数(按照列看)
print('-'*30)
#剔除缺失值
print(s.dropna())

print('-'*30)
print(df)
print(df.dropna())#在DataFrame中只要一个index行中存在缺失值,直接整条删除

#如果存在有不想删的(dataFrame中)
print(df.dropna(how='all'))#如果每行所有的值均为缺失值,才删去这一行
print(df.dropna(thresh=2))#如果每行至少有n个(这里为2)不是缺失值就保留
print(df.dropna(thresh=3))
print(df.dropna(axis=1))#剔除一整列的记录
print(df.dropna(subset=['第一列']))#如果某列有缺失值，则删除此列的该位置对应行

#填充缺失值
df = pd.read_csv('data/weather_withna.csv')
df.tail()
df.isna().sum(axis=0)
df.head()
print(df.fillna({'temp_max':20,'wind':2.5}).tail())#使用字典来填充
print(df.fillna(df[['temp_max','wind']].mean()).tail())#使用统计值来填充
print(df.ffill().tail())#用前面的相邻值进行添加front
print(df.bfill().tail())#用后面的相邻值进行填充behind
```

```python
#分组聚合
# df.groupby('分组的字段')['聚合的字段'].聚合函数()
import pandas as pd

df = pd.read_csv("data/employees.csv")
print(type(df))
print(df.dtypes)
print(df['department_id'].isna().sum())
df.dropna(how='any',subset=['department_id'],inplace = True)
df['department_id'] = df['department_id'].astype('int64')

#计算不同部门的平均薪资
df.groupby('department_id').groups#查看分组
df.groupby('department_id').get_group(20)#查看具体的某个分组数据
df2 = df.groupby('department_id')[['salary']].mean()
df2['salary'] = df2['salary'].round(2)
df2 = df2.reset_index()
df2.sort_values('salary',ascending = False)
```

```python
#计算不同部门不同岗位的平均工资
df2 = df.groupby(['department_id','job_id'])['salary'].mean()
df2 = df2.reset_index()
df2['salary'] = df2['salary'].round(1)
df2.sort_values('salary',ascending = False)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv('data/sleep.csv')
df.head(5)
df.info()
df.describe()
```

```python
#数据特征的构造
df['gender'] = df['gender'].astype('category')
df['occupation'].value_counts()
df['occupation'] = df['occupation'].astype('category')
df[['high','low']]=df['blood_pressure'].str.split("/",expand=True)#expand是否分成多列
df.head()

#睡眠质量的分箱
labels = ['低','中','高']
df['sleep_quality_level'] = pd.cut(df['sleep_quality'],bins=3,labels=labels)
age_labels = ['青少年','中年','老年']
df['age_level'] = pd.cut(df['age'],bins=3,labels = age_labels)
df.head()
```
